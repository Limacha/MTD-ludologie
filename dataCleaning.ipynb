```python
import pandas as pd

info = pd.read_csv("bg_info.csv")
designers = pd.read_csv("bg_designers.csv")

# On retire la colonne Title du second fichier (déjà présente dans le premier)
dataset = pd.concat([info, designers.drop(columns="Title")], axis=1)

dataset.shape
```

```python
# 1. Garder les colonnes utiles
cols = ["Title", "Avg rating", "Complexity", "Min time", "Max time",
        "Min players", "Max players", "Num of voters", "Type 1"]
d = dataset[cols]

d.shape
```

```python
# 2. Retirer les lignes avec valeurs manquantes
d = d.dropna(subset=["Min time", "Max time", "Min players", "Max players", "Type 1"])

d.shape
```

```python
# 3. Retirer les complexités à 0 (faux zéros)
d = d[d["Complexity"] > 0]

d.shape
```

```python
# 4. Garder les jeux avec assez de votants (notes fiables)
d = d[d["Num of voters"] >= 100]

d.shape
```

```python
# 5. Retirer les valeurs extrêmes
d = d[(d["Max time"] <= 1000) & (d["Max players"] <= 20)]

d.shape
```

```python
# 6. Créer la durée moyenne
d = d.copy()
d["Mean time"] = (d["Min time"] + d["Max time"]) / 2

d.shape
```

```python
# 7. Sauvegarder
d.to_csv("board_games_clean.csv", index=False)
d.shape
```
