Nettoyage de la base de données pour n'avoir que les colonnes utiles. On retire aussi les lignes (jeu) dont les colonnes voulues sont vides ou improbables (comme une année de sortie avant 1950).

```python
import pandas as pd
import numpy as np

dataFrame = pd.read_csv("bg_info.csv")
cleanDataFrame = dataFrame.copy()
dataFrame.shape
```

```python
# 1. Keep usefull columns
cols = ["Title", "Avg rating", "Min time", "Max time", "Num of voters", "Complexity", "Year", "Type 1"]
dataFrame = dataFrame[cols]

dataFrame.shape
```

```python
# 2. Delete lines with no values
dataFrame = dataFrame.dropna(subset=["Type 1", "Year"])
dataFrame.shape
```

```python
dataFrame["Mean time"] = (dataFrame["Min time"] + dataFrame["Max time"]) / 2
dataFrame.shape
```

```python
# 3. Delete lines with impossible Year
dataFrame = dataFrame[dataFrame["Year"] >= 1950]
dataFrame["Year"] = dataFrame["Year"].astype(int)
dataFrame.shape
```

```python
# 4. Keep lines with Complexity above 0
dataFrame = dataFrame[dataFrame["Complexity"] > 0]
dataFrame.shape
```

```python
# 5. Save cleaned dataset
dataFrame.to_csv("board_games_clean.csv", index=False)
dataFrame.shape
```
