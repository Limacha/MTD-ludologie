## 2.1 Les listes (`list`)

- **Quand ?** Quand on veut stocker une séquence **ordonnée** d’éléments, modifiable.  
- **En C/Java :** équivalent des tableaux dynamiques (`ArrayList` en Java).  
- **Particularité Python :** une liste peut contenir des éléments de types différents.


````python
notes = [12, 15, 9, 18]
print("Liste complète :", notes)
print("Premier élément :", notes[0])
print("Dernier élément :", notes[-1])
print("Sous-liste notes[1:3] :", notes[1:3])

notes[2] = 10      # modification
notes.append(20)   # ajout
print("Liste modifiée :", notes)
````

````python
ma_liste = [1, 3.5, "toto", []]
ma_liste.append(42) #ajoute à la fin
ma_liste.insert(2, 'c') #ajoute à la position spécifiée
print(ma_liste)
ma_liste.pop(2) #enlève l'élément à la position spécifiée
ma_liste
````

````python
# Parcourir une liste
fruits = ["pomme", "banane", "poire"]
for fruit in fruits:
    print(fruit)
````

On pourra créer des matrices à l'aide des listes. Mais attention, les indices commencent à 0 en python

````python
ma_matrice = [[1,2,3], [4,5,6]]
print(f"Matrice : {ma_matrice}")
print(f"Ligne 1 : {ma_matrice[0]}")

element11 = ma_matrice[0][0]
element23 = ma_matrice[-1][-1]
print(f"Premier élément (en position (1,1)) : {element11} \nDernier élément (ici, en position (2,3)) : {element23}")
somme=ma_matrice + ma_matrice #les opérations usuelles en termes de matrices ne sont pas possibles avec les listes
print(f"Ce n'est pas la somme, c'est la concaténation : {somme}")
n=len(ma_matrice) # nombre de lignes
m=len(ma_matrice[0]) # nombre de colonnes
print(f"Nb de lignes : {n} \nNb de colonnes : {m}")
````

## 2.3 Les chaînes de caractères (`str`)

## 2.6 Les compréhensions de listes

- **Quoi ?** Une compréhension de liste est une expression compacte pour construire une nouvelle liste en appliquant une fonction à chaque élément d'une autre liste ou d'un autre itérable.
- **Quand ?** Pour créer rapidement une nouvelle liste à partir d’une autre.  
- **Avantage :** syntaxe concise et lisible.

- **Syntaxe :** [expression for item in iterable if condition]


````python
nombres = [1, 2, 3, 4, 5]
carres = [x**2 for x in nombres]
pairs = [x for x in nombres if x % 2 == 0]

print("Carrés :", carres)
print("Pairs :", pairs)
````

# Module 4. Fichiers en Python

En Python, comme en C/Java, on peut lire et écrire dans des fichiers.  
Différences principales :

- On utilise la fonction **`open()`** pour ouvrir un fichier.  
- On peut choisir le **mode** :
  - `"r"` : lecture (read)
  - `"w"` : écriture (write, écrase le contenu)
  - `"a"` : ajout (append)
- Avec `with open(...) as f:`, le fichier se ferme automatiquement (bonne pratique !).

## 4.1 Lire un fichier entier

````python
# Créons un petit fichier texte pour l'exemple
with open("exemple.txt", "w", encoding="utf-8") as f:
    f.write("Python est génial\n")
    f.write("Et il est très utilisé en data science\n")

# Lecture complète
with open("exemple.txt", "r", encoding="utf-8") as f:
    contenu = f.read()

print(contenu)
````

## 4.2 Lire un fichier ligne par ligne

````python
with open("exemple.txt", "r", encoding="utf-8") as f:
    for ligne in f:
        print(">>", ligne.strip())
````

## 4.3 Écrire dans un fichier

````python
with open("resultat.txt", "w", encoding="utf-8") as f:
    f.write("Première ligne\n")
    f.write("Deuxième ligne\n")

print("Fichier 'resultat.txt' créé.")
````

NB. `write` n'accepte que des chaînes de caractères

## 4.4 Ajouter du contenu (append)

````python
with open("resultat.txt", "a", encoding="utf-8") as f:
    f.write("Troisième ligne (ajoutée)\n")

with open("resultat.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

## 4.5 Lire un fichier CSV simple
Beaucoup de données sont stockées sous forme **CSV** (Comma Separated Values).  
On peut les lire avec `open()` et `split()`.

````python
# Créons un fichier CSV simple
with open("notes.csv", "w", encoding="utf-8") as f:
    f.write("Nom,Note\n")
    f.write("Alice,14\n")
    f.write("Bob,9\n")
    f.write("Charlie,16\n")

# Lecture du CSV
with open("notes.csv", "r", encoding="utf-8") as f:
    for ligne in f:
        colonnes = ligne.strip().split(",")
        print(colonnes)
````

## 6.1 `Numpy` - Calcul scientifique rapide
NumPy est une librairie Python spécialisée dans le **calcul scientifique**.  
Elle permet de manipuler efficacement des **tableaux de données** et d’effectuer des opérations mathématiques rapides.  
C'est ce qu'on préfèrera utiliser pour manipuler des matrices.

#### Création de tableaux

````python
import numpy as np

a = np.array([4, 5, 3])
print(a)

b=np.append(a,15) #ajout de l'élément en fin de tableau
print(b)

c=np.insert(a, 2, 12) #ajout de l'élément 12 à l'adresse 2 (! on commence à compter à 0)
print(c)

d=np.delete(a,1) # supprimer l'élément à l'adresse 1
print(d)
````

#### Opérations élément par élément

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("a + b =", a + b)
print("a * b =", a * b) 
````

#### Matrices (tableaux 2D)

````python

M = np.array([[1, 2], [3, 4]])
print("Matrice M :\n", M)

# Opérations sur toute la matrice
print("Somme :", M.sum())
print("Maximum :", M.max())

print(M[0,0], M[-1,-1])
print(M + M) #les opérations usuelles sur les matrices sont possibles
print(M*M) #multiplication éléments par éléments

M2 =np.array([[1,1],[1,1]])
np.dot(M,M2) #multiplication interne
````

````python
import matplotlib.pyplot as plt

def f(x):
    return np.exp(-x) * np.cos(2*np.pi*x)

x1 = np.arange(0.0, 5.0, 0.1)
x2 = np.arange(0.0, 5.0, 0.02)

plt.subplot(211)
plt.plot(x1, f(x1), 'bo', x2, f(x2), 'k')

plt.subplot(212)
plt.plot(x2, np.cos(2*np.pi*x2), 'r--')

plt.show()
````

## 6.4 `Seaborn` – Visualisation avancée et statistique
Seaborn est une bibliothèque Python basée sur matplotlib, conçue pour créer des graphiques statistiques élégants et informatifs avec très peu de code.

````python
import seaborn as sns

# Dataset d'exemple
iris = sns.load_dataset("iris")

# 1️ Scatterplot simple
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species")
plt.title("Longueur vs Largeur des sépales")
plt.show()

# 2️ Histogramme
sns.histplot(data=iris, x="petal_length", bins=20, kde=True)
plt.title("Distribution de la longueur des pétales")
plt.show()

# 3️ Boxplot (pour visualiser la distribution et les outliers)
sns.boxplot(data=iris, x="species", y="petal_length")
plt.title("Longueur des pétales selon l'espèce")
plt.show()

# 4️ Pairplot (scatterplots multiples et distributions)
sns.pairplot(iris, hue="species")
plt.suptitle("Relations entre toutes les variables", y=1.02)
plt.show()
````
