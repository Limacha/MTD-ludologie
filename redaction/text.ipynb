# Y a-t-il une corréllation entre la popularité et complexité d'un jeu de société ?

## Introduction

Y a-t-il une corrélation entre la popularité et complexité d'un jeu de société ? Voici la question à laquelle nous allons essayer de répondre. Nous avons choisi le thème car nous sommes de gros joueurs de jeux de société. Chaque année, des prix sont accordés aux jeux les mieux réalisés, les plus populaires, etc. Nous nous sommes alors demandé ce que les jeux populaires ont en commun. Nous avons choisi d'examiner si la complexité d'un jeu est liée à sa popularité : les jeux plus complexes sont-ils, en moyenne, plus ou moins populaires que les jeux plus simples ?

Pour répondre à cette question, nous nous basons sur la base de données "Board Games info from BoardGameGeek.com", récupérée sur le site Kaggle.com (source : https://www.kaggle.com/datasets/fotopoulosvasileios/board-games-info-from-boardgamegeek-com). Cette base de données regroupe à peu près 25000 jeux ayant au moins 30 votants sur le site BoardGameGeek.com.  

La population visée est donc "les jeux listé sur BoardGameGeek.com ayant au moins 30 votants et un "Type 1". Nous reviendrons plus tard sur ce qu'est un "Type 1".

## Présentation du jeu de données

Dans la base de données que nous avons choisi, chaque ligne représente un jeu de société. On y retrouve les informations utiles suivantes : 
- Avg rating : quantitatif continu (de 0 à 10) -> note moyenne attribuée subjectivement par les votants ;
- Num of voters : quantitatif discret -> nombre de personnes qui ont voté pour ce jeu ;
- Complexity : quantitatif continu (de 1 à 5) -> valeur représentant à quel point le jeu est difficile à apprendre, à préparer, à jouer, etc, cette note est évaluée subjectivement par les joueurs ;
- Min time : quantitatif discret (en minutes) -> temps de jeu minimum ;
- Max time : quantitatif discret (en minutes) -> temps de jeu maximum ;
- Type 1 : qualitatif -> indique la catégorie de jeu parmi : abstract, children, customisable, familly, party, strategy, thematic, wargames

La base de données contient d'autres informations telles que l'année de sortie, l'auteur, le prix ou autres qui ne sont pas utiles pour notre analyse. Nous nous concentrerons alors uniquement sur les colonnes citées précédemment provenant toutes du fichier bg_info.csv, qu'on a renommé en "boardGamesRaw.csv". 

## Qualité des données

Au départ, la base de données reprend tous les jeux listés sur le site qui ont été voté par au moins 30 personnes. parmi ces 25339 lignes (jeux), certaines ne sont pas complètes. Nous nettoierons le set de données afin d'enlever les lignes qui ne contiennent pas les informations désirées. Nous enlevons notamment 47% des lignes car elles ne possèdent pas l'information "Type 1". Nous enlevons également les lignes contenant une complexité de 0 et les jeux dont la durée minimale et/ou maximale n'est pas précisée. Ce nettoyage fait que nous travaillerons avec un set de 12963 jeux. Cette suppression restreint la population à laquelle nos conclusions s'appliquent. Il est important de préciser que la plupart des votants sur le site BoardGameGeek.com sont des  passionnés, donc le set de données n'est pas représentatif de tous les joueurs.
Il contient également des noms de jeu en doublon mais nous considérons qu'il n'est pas impossible que deux jeux puissent avoir le même nom, nous les gardons alors dans le set. 

## Définition de la popularité

Ce que nous entendons lorsque nous parlons de popularité : d'après nous, un jeu populaire est un jeu qui est suffisamment connu et apprécié. Ce qui veut dire qu'il possède beaucoup de votes dont la note tend vers le maximum. 
Pour calculer la popularité d'un jeu, nous utilisons cette formule inspirée de la moyenne pondérée bayésienne : 
popularité = (V / (V + M)) x R
où : 
V = nombre de vote
R = note moyenne
M = seuil significatif de vote. C'est une valeur définie arbitrairement à 1000 qui permet de mettre en avant l'écart entre un jeu peu voté et un jeu fortement voté.
